## **EDA Mini-Dashboard Script**

**Objective:**

Create a script that:

- Loads a public dataset (or fetches data from an API).
- Automatically generates 4 different charts from the data.
- Summarizes each chart with 3–4 sentences of insight.
- Saves all charts and insights into a single output (dashboard-style).
- Handles errors gracefully using try-except.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
# Chart 1
df.groupby("Item Type")["Total Revenue"].sum().plot(kind="bar")
plt.title("Sales by Category")
plt.show()
```

```python
# Chart 2
df["Order Date"] = pd.to_datetime(df["Order Date"])
df.groupby("Order Date")["Total Revenue"].sum().plot(kind="line")
plt.title("Sales Over Time")
plt.show()
```

```python
# Chart 3
df.groupby("Region")["Total Revenue"].sum().plot(kind="bar")
plt.title("Sales by Region")
plt.show()
```

```python
# Chart 4
plt.scatter(df["Total Revenue"], df["Total Profit"])
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Sales vs Profit")
plt.show()
```
