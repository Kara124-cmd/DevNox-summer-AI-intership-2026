## Task 1

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 180, 140, 200, 220]
}

df = pd.DataFrame(data)

print(df)
```

## Bar plot

```python
plt.figure(figsize=(8, 5))

plt.bar(df["Month"], df["Sales"])

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

## Line Plot

```python
plt.figure(figsize=(8, 5))

plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

## Pie Chart

```python
plt.figure(figsize=(7, 7))

plt.pie(
    df["Sales"],
    labels=df["Month"],
    autopct="%1.1f%%"
)

plt.title("Sales Distribution")

plt.show()
```

#  **Task 2**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create sample numeric dataset
data = {
    "Hours_Studied": [2, 3, 4, 5, 6, 7, 8, 9],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 95],
    "Assignments": [4, 5, 5, 6, 7, 7, 8, 9],
    "Marks": [45, 50, 55, 60, 68, 72, 80, 88]
}

df = pd.DataFrame(data)

# Calculate correlation
correlation = df.corr()

print(correlation)
```

```python
# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

## **Task 4**

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

# 1. API URL
url = "https://jsonplaceholder.typicode.com/posts"

# 2. Fetch data from API
response = requests.get(url)

# 3. Check if request was successful
if response.status_code == 200:

    # 4. Parse JSON response
    data = response.json()

    # 5. Convert JSON data into DataFrame
    df = pd.DataFrame(data)

    # 6. Count posts created by each user
    user_posts = df["userId"].value_counts().sort_index()

    # 7. Create bar chart
    user_posts.plot(
        kind="bar",
        figsize=(10, 6),
        title="Number of Posts by User"
    )

    # 8. Add labels
    plt.xlabel("User ID")
    plt.ylabel("Number of Posts")
    plt.xticks(rotation=0)

    # 9. Display chart
    plt.show()

else:
    print("Failed to fetch data.")
    print("Status code:", response.status_code)
```

## **Task 5**

```python
import pandas as pd
import matplotlib.pyplot as plt

# dataset

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [120, 150, 180, 140, 200]
}

df = pd.DataFrame(data)



# Bar Chart
def plot_bar(df, x, y, title="Bar Chart"):
    plt.figure(figsize=(8, 5))
    plt.bar(df[x], df[y])
    plt.xlabel(x)
    plt.ylabel(y)
    plt.title(title)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# Bar chart
plot_bar(
    df,
    x="Month",
    y="Sales",
    title="Monthly Sales - Bar Chart"
)
```

```python
#  Line Chart
def plot_line(df, x, y, title="Line Chart"):
    plt.figure(figsize=(8, 5))
    plt.plot(df[x], df[y], marker="o")
    plt.xlabel(x)
    plt.ylabel(y)
    plt.title(title)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# Line chart
plot_line(
    df,
    x="Month",
    y="Sales",
    title="Monthly Sales - Line Chart"
)
```

```python
# Histogram
def plot_histogram(df, column, bins=10, title="Histogram"):
    plt.figure(figsize=(8, 5))
    plt.hist(df[column], bins=bins)
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title(title)
    plt.tight_layout()
    plt.show()

# Histogram
plot_histogram(
    df,
    column="Sales",
    bins=5,
    title="Sales Distribution"
)
```
