# **OWASP Top 10**

The OWASP Top 10 is a list of the most important and common web application security risks identified by the Open Worldwide Application Security Project (OWASP).

Assuming your course is using the current OWASP Top 10:2025, here is a one-sentence summary of each:

1. A01 – Broken Access Control: Attackers can access or modify data and functions that they should not be allowed to access.
2. A02 – Security Misconfiguration: Incorrect or insecure settings in applications, servers, databases, or cloud services can create security vulnerabilities.
3. A03 – Software Supply Chain Failures: Vulnerabilities or malicious components in third-party libraries, dependencies, tools, or software development processes can compromise an application.
4. A04 – Cryptographic Failures: Sensitive information can be exposed when encryption or other cryptographic protections are missing, weak, or incorrectly implemented.
5. A05 – Injection: Attackers can insert malicious commands or data into an application so that the application executes unintended instructions.
6. A06 – Insecure Design: Security weaknesses can occur when security requirements and protections are not properly considered during the application's design.
7. A07 – Authentication Failures: Weak login, password, session, or identity-management mechanisms can allow attackers to impersonate legitimate users.
8. A08 – Software or Data Integrity Failures: Applications can be compromised when they trust software, updates, plugins, or data without properly verifying their integrity.
9. A09 – Security Logging & Alerting Failures: Poor logging and monitoring can prevent organizations from detecting, investigating, and responding to attacks effectively.
10. A10 – Mishandling of Exceptional Conditions: Improper handling of errors, unexpected inputs, or unusual system conditions can cause security weaknesses or expose sensitive information.

`Simple way to remember
`
   -  Access → Configuration → Supply Chain → Cryptography → Injection
       → Design → Authentication → Integrity → Logging → Exceptions
