**`Tokenization`**
   - Tokenization means breaking text into smaller pieces called tokens.
   - Tokens can be words, subwords, or characters.
   - For example, the sentence "I love programming" can be tokenized into ["I", "love", "programming"].
   - **`Stemming`**
   - Stemming is the process of reducing words to their base or root form.
   - For example, the words "running", "runner", and "ran" can be
   - reduced to the root word "run".
   - **`Lemmatization`**
   - Lemmatization is similar to stemming, but it reduces words to their base or dictionary form (lemma).
   - For example, the words "running", "runner", and "ran" can be reduced to the lemma "run".
   - **`Stop Words`**
   - Stop words are common words that are often removed from text during preprocessing because they do not carry significant meaning.
   - Examples of stop words include "the", "is", "in", "and", etc.
   - **`Bag of Words (BoW)`**
   - Bag of Words is a representation of text that describes the occurrence of words within a document
   - - It ignores grammar and word order but keeps track of the frequency of each word. 

## Manual Tokenization:
  Manual tokenization involves breaking down text into tokens without using any pre-built libraries or tools. This can be done using simple string manipulation techniques in programming languages like Python. For example, you can split a sentence into words using the `split()` method.
  

```python
#  Manual Tokenization

import re

paragraph = """
Artificial Intelligence is changing the world.
Natural Language Processing allows computers to understand human language.
"""

# Manual tokenization
tokens = re.findall(r"\b\w+\b|[^\w\s]", paragraph)

print("Manual Tokens:")
print(tokens)

print("\nNumber of tokens:", len(tokens))
```

## Hugging Face Tokenizer:
  

```python
from transformers import AutoTokenizer

# Load a pretrained tokenizer
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")

# Tokenize the same paragraph
hf_tokens = tokenizer.tokenize(paragraph)

print("Hugging Face Tokens:")
print(hf_tokens)

print("\nNumber of tokens:", len(hf_tokens))
```
