```python
import pandas as pd

test_df = pd.DataFrame({
    "feature1": [1, 2, 2, 3],
    "feature2": [4, 5, 5, 6],
    "target": [0, 1, 1, 0]
})

cleaned_df = clean_data(test_df)

cleaned_df = cleaned_df.drop_duplicates(ignore_index=True)
assert cleaned_df.duplicated().sum() == 0
```

```python
def clean_data(df):
    df = df.dropna()
    return df
```

only removes missing values. It does not remove duplicate rows.

```python
# Fix the Bug

def clean_data(df):
    df = df.drop_duplicates()
    df = df.dropna()
    return df
```

```python
# Run the Test Again
cleaned_df = clean_data(test_df)

assert cleaned_df.duplicated().sum() == 0

print("Bug fixed: duplicate rows are removed.")
```
