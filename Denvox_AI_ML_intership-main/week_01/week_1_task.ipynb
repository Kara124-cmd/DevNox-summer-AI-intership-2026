```python
# Program to check if a number is prime and if a string is a palindrome

# Check prime number
num = int(input("Enter a number: "))

if num <= 1:
    print("Not Prime")
else:
    for i in range(2,  num):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")

# Check palindrome
str = input("Enter a string: ")
str = str.lower()

reversed_str = str[::-1]

if str == reversed_str:
    print("Palindrome")
else:
    print("Not palindrome:")
```

```python
# task 3

# Print the Fibonnacci series using loop

def fib(num):
    a , b = 0 , 1
    while a < num:
        print(a, end=' ')
        a, b = b , a+b
    print()

fib(5)
```

```python
def fibonacci(n):
  if n <= 1:
    return n
  else:
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(6))
```

```python
#  task 4
# Build a small function library: temperature converter + simple area calculator (circle, rectangle).

def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius+32)*9/5
    return fahrenheit

def fahrenheit_to_celsius(fahrenheit):
    celsius = (fahrenheit-32)*5/9
    return celsius

def circle_area(radius, pi=3.1416):
    area = pi * radius * radius
    return area

def rect_area(l , w):
    area = l*w
    return area

def square_area(side):
    area = side * side
    return area

# used loop to call the function one by one
while True:

    print("\n-------Function Library-------")
    print("1. Celsius to fahrenheit")
    print("2. Faghrenheit to celsius")
    print("3. Circle Area")
    print("4. Rectanglar Area")
    print("5. Square Area")
    print("6. Exit")

    choice = int(input("Enter the choice:"))

    if choice == 1:

        c = float(input("Enter Celsius"))
        print("Fahrenheit =", celsius_to_fahrenheit(c))

    elif choice == 2:
        f = float(input("Enter Fahrenheit"))
        print("Fahrenheit  = ", fahrenheit_to_celsius(f))

    elif choice == 3:
        r = float(input("Enter the Radius : "))
        print("Circle Area = ", circle_area(r))

    elif choice == 4:
        l = float(input())
        w = float(input())
        print("Rectangular Area = ", rect_area(l,w))

    elif choice == 5:
        square = float(input())
        print("Square Area = ", square_area(square))

    elif choice == 6:
        print("Program Closed")
        break

    else:
        print("Invalid Choice!")
```

```python
#  Task 5

class shape:
    def __init__(self):
        print("Shape Object Created")

    def area(self):
        print("Area is not defined.")

class circle(shape):
    def __init__(self, radius):
        super().__init__()
        self.radius = radius

    def area(self):
        circle_area = 3.1416 * self.radius * self.radius
        print("Circle Area : ", circle_area)


class Rectangle(shape):
    def __init__(self, l, w):
        super().__init__()
        self.l = l
        self.w = w

    def area(self):
        area = self.l * self.w
        print("Rectangle Area = ". area)


# create object

circle = circle(5)
circle.area()

Rectangle = Rectangle(10, 6)
Rectangle.area

      
```

```python
#  Reverse a String (Without Using Built-in Reverse Function)
def reverse_string(text):

    reverse = ""

    for char in text:
        reverse = char + reverse

    return reverse


word = input("Enter a String: ")

print("Reversed String:", reverse_string(word))
```

```python
#  Check if Two Strings are Anagrams
def is_anagram(str1, str2):

    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()

    return sorted(str1) == sorted(str2)


first = input("Enter First String: ")
second = input("Enter Second String: ")

if is_anagram(first, second):
    print("Both strings are Anagrams.")
else:
    print("Not Anagrams.")
```

```python
# : Find the Second Largest Number in a List
def second_largest(numbers):

    unique_numbers = list(set(numbers))
    unique_numbers.sort()

    return unique_numbers[-2]


numbers = [10, 5, 80, 30, 70, 80]

print("Second Largest Number:", second_largest(numbers))
```

```python
# Student Class with Grade Method
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def compute_grade(self):

        if self.marks >= 90:
            return "A+"

        elif self.marks >= 80:
            return "A"

        elif self.marks >= 70:
            return "B"

        elif self.marks >= 60:
            return "C"

        elif self.marks >= 50:
            return "D"

        else:
            return "Fail"


student1 = Student("Basit", 88)

print("Name :", student1.name)
print("Marks:", student1.marks)
print("Grade:", student1.compute_grade())
```

# Task 6:

```python
#  Simple To-Do List App
tasks = []

def add_task():

    task = input("Enter Task: ")

    tasks.append(task)

    print("Task Added Successfully.")


def view_tasks():

    if len(tasks) == 0:
        print("No Tasks Found.")

    else:
        print("\nYour Tasks:")

        for i, task in enumerate(tasks, start=1):
            print(i, ".", task)


def remove_task():

    view_tasks()

    if len(tasks) > 0:

        index = int(input("Enter Task Number to Remove: "))

        if 1 <= index <= len(tasks):

            tasks.pop(index - 1)

            print("Task Removed.")

        else:

            print("Invalid Task Number.")


while True:

    print("\n====== TO-DO LIST ======")
    print("1. Add Task")
    print("2. View Tasks")
    print("3. Remove Task")
    print("4. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_task()

    elif choice == "2":
        view_tasks()

    elif choice == "3":
        remove_task()

    elif choice == "4":
        print("Program Closed.")
        break

    else:
        print("Invalid Choice.")
```
