`Build a perceptron manually (weights, bias, step function) and test it on a simple case.`

**Perceptron**: 
- A perceptron is a simple linear binary classifier that makes its predictions based on a linear predictor function combining a set of weights with the feature vector. It consists of weights, a bias term, and an activation function (step function).
- step function:
- 
  - If the weighted sum of inputs is greater than or equal to a threshold (usually 0), the output is 1 (positive class).
  - If the weighted sum of inputs is less than the threshold, the output is 0 (negative class).
  


```python
import numpy as np

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])

weights = np.array([1, 1])
bias = -1.5


def step_function(z):
    return 1 if z >= 0 else 0


def perceptron(inputs, weights, bias):
    z = np.dot(inputs, weights) + bias
    return step_function(z)


for inputs, actual in zip(X, y):

    prediction = perceptron(inputs, weights, bias)

    print(
        "Input:", inputs,
        "Actual:", actual,
        "Predicted:", prediction
    )
```
