```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Messy_Employee_dataset.csv")

original_rows = len(df)
```

```python
# 1 Clean column names

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
# 2 Remove duplicates

duplicates_removed = df.duplicated().sum()

df = df.drop_duplicates()
```

```python
#  Handle missing Age

age_missing = df["age"].isnull().sum()

df["age"] = df["age"].fillna(df["age"].median())

#  Handle missing Gender

if "gender" in df.columns:
    gender_missing = df["gender"].isnull().sum()
else:
    gender_missing = len(df)
    df["gender"] = pd.NA

df["gender"] = df["gender"].fillna("unknown")
```

```python
# Standardize Gender

df["gender"] = df["gender"].str.strip().str.lower()

df["gender"] = df["gender"].replace({
    "m": "male",
    "f": "female"
})

# Standardize City

if "city" in df.columns:
    df["city"] = df["city"].astype(str).str.strip().str.title()
else:
    df["city"] = pd.NA
```

```python
# Clean Salary

df["salary"] = (
    df["salary"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)

# 9. Save cleaned data

df.to_csv("cleaned_data.csv", index=False)

```

```python
# 10. Create report

report = f"""
DATA CLEANING REPORT
--------------------

Original rows: {original_rows}
Final rows: {len(df)}

Duplicates removed: {duplicates_removed}

Missing Age values filled: {age_missing}
Missing Gender values filled: {gender_missing}

Formatting fixes:
- Column names standardized
- Gender values standardized
- City names standardized
- Salary commas removed
- Salary converted to numeric

Output file:
cleaned_data.csv
"""

with open("cleaning_report.txt", "w") as file:
    file.write(report)

print("Data cleaning completed!")
print("Cleaned file: cleaned_data.csv")
print("Report: cleaning_report.txt")
```
