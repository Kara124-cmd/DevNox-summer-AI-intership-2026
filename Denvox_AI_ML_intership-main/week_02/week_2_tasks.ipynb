# Task 3:
Practice Pandas: load a dataset, filter rows, group by a column, and handle missing values.

```python
import pandas as pd

# Load dataset
df = pd.read_csv("students.csv")

# Display data
print(df.head())

# Filter students with marks greater than 80
print(df[df['Marks'] > 80])

# Average marks by department
print(df.groupby('Department')['Marks'].mean())

# Check missing values
print(df.isnull().sum())

# Replace missing values with the mean
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

# Display updated data
print(df)
```

# Task 4:
Merge two small datasets (dictionaries or DataFrames) and produce a summary.

```python
import pandas as pd

students = {
    'ID':[1,2,3,4],
    'Name':['Ali','Sara','Ahmed','Ayesha']
}

marks = {
    'ID':[1,2,3,4],
    'Marks':[85,90,78,95]
}

df_students = pd.DataFrame(students)
df_marks = pd.DataFrame(marks)
```

```python
# print the students dataframe
print(df_students)
```

```python
print(df_marks)
```

```python
# merged the dataframe into single dataframe
merged_df = pd.merge(df_students, df_marks, on="ID")
print(merged_df)
```

# types of merge function

```python
# 1 inner join
merged_inner = pd.merge(df_students, df_marks, on="ID", how='inner')
print(merged_inner)

print("----")
# left join
merged_left = pd.merge(df_students, df_marks, on="ID", how='left')
print(merged_left)

print("----")
# right join 
merged_right = pd.merge(df_students, df_marks, on="ID", how='right')
print(merged_right)

print("----")
merged_outer = pd.merge(df_students, df_marks, on='ID', how='outer')
print(merged_outer)
```

```python
# Average Marks
print(merged_df['Marks'].mean())

print("---")
# Highest Marks
print(merged_df['Marks'].max())

print("---")
# Lowest Marks
print(merged_df['Marks'].min())

print("---")
# Total Marks
print(merged_df['Marks'].sum())
```

```python
# combine code 
import pandas as pd

students = {
    'ID':[1,2,3,4],
    'Name':['Ali','Sara','Ahmed','Ayesha']
}

marks = {
    'ID':[1,2,3,4],
    'Marks':[85,90,78,95]
}

df_students = pd.DataFrame(students)
df_marks = pd.DataFrame(marks)

merged_df = pd.merge(df_students, df_marks, on='ID')

print(merged_df)

print("\nAverage Marks:", merged_df['Marks'].mean())
print("Highest Marks:", merged_df['Marks'].max())
print("Lowest Marks:", merged_df['Marks'].min())
print("Total Marks:", merged_df['Marks'].sum())

print("\nSummary Statistics:")
print(merged_df.describe())
```
